Step-1 Tokenization: Convert words into tokens (mathematical numbers), it also varies from model to model

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("google/gemma-3-1b-it")
tokenizer("Hey There!, My name is aditya")
```

Next step is to generate the vector embeddings corresponding to the input tokens

Now we're just going through the working of LLMs

```python
input_tokens = tokenizer("Write a python code of adding two numbers", return_tensors="pt")
input_tokens
```

```python
import torch
from transformers import AutoModelForCausalLM
model = AutoModelForCausalLM.from_pretrained("google/gemma-3-1b-it", torch_dtype=torch.bfloat16)
```

```python
out = model(input_ids=input_tokens["input_ids"])
out ##It is basically the labels or the probabilities of the next words
```

```python
gen_out = model.generate(input_ids=input_tokens["input_ids"], max_new_tokens=100)
gen_out
```

```python
tokenizer.batch_decode(gen_out)
```
